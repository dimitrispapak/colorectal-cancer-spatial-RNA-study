```python
import requests
import pandas as pd
from io import StringIO

def get_cell_markers(cell_type):
    url = "http://biocc.hrbmu.edu.cn/CellMarker/download/all_cell_markers.txt"
    response = requests.get(url)
    #data = pd.read_csv(pd.compat.StringIO(response.text), sep='\t')
    data = pd.read_csv(StringIO(response.text), sep='\t')
    print(data)
    # Filter for the specific cell type
    cell_type_data = data[data['cell_type'].str.contains(cell_type, case=False, na=False)]
    
    # Extract unique genes
    genes = set(cell_type_data['marker'].str.split(',').sum())
    
    return {
        'genes': genes,
        'species': cell_type_data['species'].unique().tolist(),
        'tissues': cell_type_data['tissue'].unique().tolist()
    }

# Example usage
nk_cell_markers = get_cell_markers('Natural Killer')
print(f"NK cell markers: {nk_cell_markers['genes']}")
print(f"Species: {nk_cell_markers['species']}")
print(f"Tissues: {nk_cell_markers['tissues']}")
```
